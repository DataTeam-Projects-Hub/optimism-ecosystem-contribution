```python
# Store CPI results
cpi_results = {}

# Add token_house CPI (for th_vp only)
cpi_results["token_house"] = round((data["th_vp"] ** 2).sum(), 2)

# Iterate and calculate each influence + CPI
for period in influence_periods:
    label = f"influence_{period['label']}"
    weights = period["influences"]

    # Weighted sum of relevant columns
    data[label] = round(
        sum(data[col] * (weights[col] / 100) for col in weights),
        6
    )

    # Calculate CPI
    cpi = round((data[label] ** 2).sum(), 2)
    cpi_results[label] = cpi

```

```python
# Save CPI results to CSV
cpi_df = pd.DataFrame(cpi_results.items(), columns=["influence_period", "CPI"])
cpi_df.to_csv("../Data/influence_periods_cpi.csv", index=False)
```

```python
# Plot
plt.figure(figsize=(12, 6))
plt.bar(cpi_df['influence_period'], cpi_df['CPI'], color='skyblue', edgecolor='black')

# Titles and labels
plt.title("Concentration of Power Index (CPI) by Influence Period", fontsize=14)
plt.xlabel("Influence Period", fontsize=12)
plt.ylabel("CPI (Sum of Squares)", fontsize=12)
plt.xticks(rotation=45, ha='right')

# Annotate each bar with its CPI value
for idx, value in enumerate(cpi_df['CPI']):
    plt.text(idx, value + 0.002, f"{value:.2f}", ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```

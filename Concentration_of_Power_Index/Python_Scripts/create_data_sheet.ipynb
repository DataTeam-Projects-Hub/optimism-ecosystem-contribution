```python
# Filter grants data by season 
grants_season3 = grants[grants['season'] == 3]

grants_season4 = grants[grants['season'] == 4]

acc_season5 = acc[acc['season'] == 5]
grants_season5 = grants[grants['season'] == 5]
grants_mm_season5 = grants_mm[grants_mm['season'] == 5]
dab_season5 = dab[dab["season"] == 5]
coc_season5 = coc[coc["season"] == 5]

acc_season5 = acc[acc['season'] == 6]
grants_season6 = grants[grants['season'] == 6]
grants_mm_season6 = grants_mm[grants_mm['season'] == 6]
dab_season6 = dab[dab["season"] == 6]
coc_season6 = coc[coc["season"] == 6]

acc_season7 = acc[acc['season'] == 7]
grants_season7 = grants[grants['season'] == 7]
grants_op_season7 = grants_op[grants_op['season'] == 7]
dab_season7 = dab[dab["season"] == 7]
mmc_season7 = mmc[mmc["season"] == 7]
```

```python
# Helper function to add missing members
def add_missing_members(data, new_addresses):
    missing_addresses = set(new_addresses['address'].str.lower()) - set(data['delegate'].str.lower())
    missing_df = pd.DataFrame({
        'delegate': list(missing_addresses),
        'voting_power': 0
    })
    data = pd.concat([data, missing_df], ignore_index=True)
    return data
```

```python
# Load the daily delegate data
data = pd.read_csv("../Data/op_delegates.csv")

# Add missing delegates from various rounds and councils (unchanged)
data = add_missing_members(data, citizens_round2)
data = add_missing_members(data, citizens_round3)
data = add_missing_members(data, citizens_round4)
data = add_missing_members(data, citizens_round5)
data = add_missing_members(data, citizens_round6)
data = add_missing_members(data, citizens_round7)
data = add_missing_members(data, grants_season3)
data = add_missing_members(data, grants_season4)
data = add_missing_members(data, grants_season5)
data = add_missing_members(data, grants_mm_season5)
data = add_missing_members(data, dab_season5)
data = add_missing_members(data, coc_season5)
data = add_missing_members(data, grants_season6)
data = add_missing_members(data, grants_mm_season6)
data = add_missing_members(data, dab_season6)
data = add_missing_members(data, coc_season6)
data = add_missing_members(data, grants_season7)
data = add_missing_members(data, grants_op_season7)
data = add_missing_members(data, dab_season7)
data = add_missing_members(data, mmc_season7)

# Add the columns
data['ch_member_r2'] = None
data['ch_member_r3'] = None
data['ch_member_r4'] = None
data['ch_member_r5'] = None
data['ch_member_r6'] = None
data['ch_member_r7'] = None

data['gc_member_s3'] = None
data['gc_member_s4'] = None

data['gc_member_s5'] = None
data['gc_member_mm_s5'] = None
data['sc_member_s5'] = None
data['coc_member_s5'] = None
data['dab_member_s5'] = None

data['gc_member_s6'] = None
data['gc_member_mm_s6'] = None
data['sc_member_s6'] = None
data['coc_member_s6'] = None
data['dab_member_s6'] = None

data['gc_member_s7'] = None
data['gc_member_op_s7'] = None
data['sc_member_s7'] = None
data['dab_member_s7'] = None
data['mmc_member_s7'] = None

# Create membership columns for existing delegates only if the round/council is active on the file_date
data['ch_member_r2'] = data['delegate'].apply(
    lambda x: 1 if x.lower() in citizens_round2['address'].str.lower().values else 0)

data['ch_member_r3'] = data['delegate'].apply(
    lambda x: 1 if x.lower() in citizens_round3['address'].str.lower().values else 0)

data['ch_member_r4'] = data['delegate'].apply(
    lambda x: 1 if x.lower() in citizens_round4['address'].str.lower().values else 0)

data['ch_member_r5'] = data['delegate'].apply(
    lambda x: 1 if x.lower() in citizens_round5['address'].str.lower().values else 0)

data['ch_member_r6'] = data['delegate'].apply(
    lambda x: 1 if x.lower() in citizens_round6['address'].str.lower().values else 0)

data['ch_member_r7'] = data['delegate'].apply(
    lambda x: 1 if x.lower() in citizens_round7['address'].str.lower().values else 0)

data['gc_member_s3'] = data['delegate'].apply(
    lambda x: 1 if x.lower() in grants_season3['address'].str.lower().values else 0)

data['gc_member_s4'] = data['delegate'].apply(
    lambda x: 1 if x.lower() in grants_season4['address'].str.lower().values else 0)

data['gc_member_s5'] = data['delegate'].apply(
    lambda x: 1 if x.lower() in grants_season5['address'].str.lower().values else 0)

data['gc_member_mm_s5'] = data['delegate'].apply(
    lambda x: 1 if x.lower() in grants_mm_season5['address'].str.lower().values else 0)

data['dab_member_s5'] = data['delegate'].apply(
    lambda x: 1 if x.lower() in dab_season5['address'].str.lower().values else 0)

data['coc_member_s5'] = data['delegate'].apply(
    lambda x: 1 if x.lower() in coc_season5['address'].str.lower().values else 0)

data['gc_member_s6'] = data['delegate'].apply(
    lambda x: 1 if x.lower() in grants_season6['address'].str.lower().values else 0)

data['gc_member_mm_s6'] = data['delegate'].apply(
    lambda x: 1 if x.lower() in grants_mm_season6['address'].str.lower().values else 0)

data['dab_member_s6'] = data['delegate'].apply(
    lambda x: 1 if x.lower() in dab_season6['address'].str.lower().values else 0)

data['coc_member_s6'] = data['delegate'].apply(
    lambda x: 1 if x.lower() in coc_season6['address'].str.lower().values else 0)

data['gc_member_s7'] = data['delegate'].apply(
    lambda x: 1 if x.lower() in grants_season7['address'].str.lower().values else 0)

data['gc_member_op_s7'] = data['delegate'].apply(
    lambda x: 1 if x.lower() in grants_op_season7['address'].str.lower().values else 0)

data['dab_member_s7'] = data['delegate'].apply(
    lambda x: 1 if x.lower() in dab_season7['address'].str.lower().values else 0)

data['mmc_member_s7'] = data['delegate'].apply(
    lambda x: 1 if x.lower() in mmc_season7['address'].str.lower().values else 0)

pd.set_option('future.no_silent_downcasting', True)
data.fillna(0, inplace=True)
```

```python
api_key = "WPWyWJJZEPTxvOwQmpcHbfbHoqXHaWxv"  # Replace with your Dune API key
query_id = 5351295  # Replace with your query ID

# Initialize the DuneClient
dune = DuneClient(api_key)

# Function to fetch Dune query results
def fetch_dune_query_results(query_id):
    try:
        query_result = dune.get_latest_result(query_id)
        return query_result.result.rows  # Access the rows in the response
    except Exception as e:
        print(f"Error fetching query results: {e}")
        return None

# Function to write results to CSV
def write_to_csv(data, file_name):
    try:
        with open(file_name, mode='w', newline='', encoding='utf-8') as file:
            writer = csv.writer(file)
            
            if len(data) > 0:
                headers = data[0].keys()  # Get keys from the first row
                writer.writerow(headers)
                
                for row in data:
                    writer.writerow(row.values())
        print(f"Data written to {file_name}")
    except Exception as e:
        print(f"Error writing to CSV: {e}")

# Main code to fetch and save results
if __name__ == "__main__":
    result = fetch_dune_query_results(query_id)
    if result:
        file_name = "../Data/optimism_delegates_ens.csv"
        write_to_csv(result, file_name)
```
